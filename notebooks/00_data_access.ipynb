# 00 Data access

Loads the bond-month panel from the group database (`ML2_DB_URL` in `.env`) or, if that is empty, from `data/processed/bond_month_panel.parquet`. See `docs/DATA_ACCESS.md`.

```python
import pandas as pd

from ml2.config import load_config
from ml2.data import load_column_dictionary, load_panel

cfg = load_config()
df = load_panel()
df.shape
```

```python
load_column_dictionary()
```

```python
# Rows per partition and grade. Never look at the test partition's targets before the final table.
pd.crosstab(df["partition"], df["grade_bucket"], margins=True)
```
