# 00 Data access

Checks that your copy of the panel is the shared version, loads it and shows what is in it. By default it reads
`data/processed/bond_month_panel.parquet`; if `ML2_DB_URL` is set in `.env` it reads the group database instead.
See `docs/DATA_ACCESS.md`. If every cell runs, your setup is done.

```python
import pandas as pd

from ml2.config import load_config
from ml2.data import check_panel, load_column_dictionary, load_panel

cfg = load_config()
info = check_panel()  # raises if the parquet file is missing or not the shared version (database users: skip)
print(f"Built {info['built_at_utc']}: {info['rows']:,} rows, {info['bonds']:,} bonds, {info['issuers']} issuers, "
      f"{info['snapshots']} month-ends ({info['first_snapshot']} to {info['last_snapshot']})")
```

```python
df = load_panel()
df.shape
```

```python
load_column_dictionary()
```

```python
# Rows per partition and grade. load_panel leaves test out: it is opened once, in 99_final.
print(f"test: {info['rows_by_partition']['test']:,} rows (opened once, in 99_final)")
pd.crosstab(df["partition"], df["grade_bucket"], margins=True)
```

```python
train = df[df["partition"] == "train"]
train.groupby(["grade_bucket", "maturity_bucket"])[cfg["problem"]["target"]].describe().round(1)
```
